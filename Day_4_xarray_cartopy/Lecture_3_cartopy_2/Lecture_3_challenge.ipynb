### Skywatch data

Did you know that CU Boulder has its very own weather station called Skywatch? You can use data from Skywatch to check how much our own instruments measured and compare with the national snowfall data. 

To get Skywatch data:

- Go to [Skywatch Archives](https://skywatch.colorado.edu/php/archives.php)
- Note down the weather station information (latitude, longitude, altitude)
- In the list of instruments, use the `Precipitation Gauge`
- Be sure to adjust the date before/after the national snowfall data file date since the latter is a 24-hour accumulation  
- Click `Submit`
- You should now see a plot displayed of precipitation (inches) vs. time
- Let's instead grab the data file 
  - Click on the `data file` and you should see a data file displayed

It looks like the data is in strings, has semicolons, tabs and a bunch of weird formatting that's hard to work with. So let's instead try to use `pandas` to read in the file

```python
import pandas as pd
skywatch_df = pd.read_csv(
    data_file,
    sep="\t",          # The file uses tabs between columns
    comment=";",       # Skips all lines starting with ';'
    names=["Time_UTC", "Precipitation_inches"] # Gives columns clean names
)
```
