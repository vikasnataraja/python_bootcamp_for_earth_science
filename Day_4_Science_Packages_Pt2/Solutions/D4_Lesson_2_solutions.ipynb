## 5.4: Cartopy

#### To make a base map look more professional, we are going to import a new package called `cartopy`

```python
import cartopy.crs as ccrs
import cartopy.feature as cfeature
```

Cartopy allows you to plot a given spatial dataset on an actual map of the world (or specific region) so that it looks nice. It will work with all sorts of datasets and relies on the plotting power of matplotlib

Let's start by creating a blank map of the world and add coastlines of land regions

```python
# To start, we will plot a blank map with the PlateCarree projection...
ax = plt.axes(projection=ccrs.PlateCarree())
ax.coastlines() # add coastlines to the map

# at it's simplest, cartopy allows you to plot a blank map of the globe with no data
plt.show()
plt.close()
```

*Instructor hint: note that `plt.axes(projection=...)` implicitly creates a new figure - students coming from plain matplotlib sometimes expect they need `plt.figure()` first and get confused when skipping it doesn't cause an error.*

Cartopy has several in-built features. For instance, land, ocean, gridlines, county borders, etc. Take a look at some options for [features](https://scitools.org.uk/cartopy/docs/v0.14/matplotlib/feature_interface.html).

```python
# aside from coastlines, there are various other features you can add to maps!
ax = plt.axes(projection=ccrs.PlateCarree())

ax.coastlines() # Coastlines
ax.add_feature(cfeature.LAND) # Land
ax.add_feature(cfeature.BORDERS) # National Border
plt.show()
plt.close()
```

#### As mentioned earlier, Cartopy has a number of different projections you can use.

```python
# plot a blank map with the Robinson projection
ax = plt.axes(projection=ccrs.Robinson())
ax.coastlines()
plt.show()
plt.close()
```

```python
# Cartopy also has built in base maps, that allow you to plot what looks like the Earth from space
ax = plt.axes(projection=ccrs.Robinson())
ax.stock_img() # satellite stock image of earth
ax.coastlines() # Coastlines
ax.add_feature(cfeature.BORDERS) # National borders
plt.show()
plt.close()
```

<font color="#0F766E">Add in ocean, land, and coastlines. Do you need all 3 features?</font>

```python
# SouthPolarStereo
ax = plt.axes(projection=ccrs.SouthPolarStereo(central_longitude=0))
ax.add_feature(cfeature.LAND)
ax.add_feature(cfeature.OCEAN)
ax.coastlines()
plt.show()
plt.close()
```

**Answer:** SouthPolarStereo is useful any time the region of interest is centered on Antarctica or the Southern Ocean - PlateCarree and Robinson badly stretch and distort the poles, while this projection keeps the pole undistorted at the center of the plot. Adding both `LAND` and `OCEAN` makes the coastline much easier to read than `coastlines()` alone.

```python
# Orthographic
ax = plt.axes(projection=ccrs.Orthographic(central_longitude=-75, central_latitude=40))
ax.add_feature(cfeature.LAND)
ax.add_feature(cfeature.OCEAN)
ax.coastlines()
plt.show()
plt.close()
```

**Answer:** Orthographic mimics the view of Earth as seen from space, centered on whatever longitude/latitude you choose. It's great for presentations or for illustrating a single hemisphere at a time, but it can only ever show half the globe at once and heavily distorts anything near its edge.

```python
# Projection of choice: Mollweide
ax = plt.axes(projection=ccrs.Mollweide(central_longitude=150))
ax.add_feature(cfeature.LAND)
ax.add_feature(cfeature.OCEAN)
ax.coastlines()
plt.show()
plt.close()
```

**Answer:** Mollweide is an equal-area projection, so it's a good choice whenever the relative *area* of features matters for a visual comparison (e.g., comparing the extent of two biomes), unlike Robinson, which only compromises between area and shape distortion rather than preserving either one.

*Instructor hint: no flat projection preserves area, shape, distance, and direction all at once - it's worth asking students which of those properties matters most for their own research data before they pick a default projection going forward.*

### Cartopy also allows you to zoom in and out

You can subset data to specific regions on maps themselves, meaning you don't necessarily need to subset the data itself by latitude and longitude, you are simply changing the view point.

```python
ax = plt.axes(projection=ccrs.Robinson(central_longitude=180))
# you can change the central longitude of the map

ax.coastlines()

# we earlier subset to latitudes -65.5:65.5 and longitudes from 109.5:288.5
# We can use ax.set_extent([x0,x1,y0,y1]) to replicate this without limiting our data!

ax.set_extent([109.5,288.5,-65.5, 65.5])
plt.show()
plt.close()
```

*Instructor hint: emphasize that `ax.set_extent()` only changes the *view* - the full global grid is still being processed under the hood. For large datasets, actually slicing the data (as in 5.1) is more efficient; `set_extent` is best reserved for quick visual exploration rather than performance-sensitive work.*
