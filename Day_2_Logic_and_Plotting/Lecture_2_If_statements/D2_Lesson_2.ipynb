### 1.2 `elif` and `else`

When you need **more than two outcomes**, use `if`, then zero or more `elif` (else if) branches, then an optional `else` for everything left over. The series must start with `if`.


<font color="#0F766E">**Task:** Look at the code below, but **do not run it yet**. What will print? Write your prediction above the code cell.</font>


*(Prediction for a=34, b=33:)*


```python
a = 34
b = 33

if a > b:
    print('a is greater than b')
elif a < b:
    print('a is less than b')
else:
    print('a is equal to b')
```

**Wind speed categories** (same thresholds as this afternoon's challenge):

- `calm` if speed &lt; 5 m/s
- `moderate` if 5 ≤ speed ≤ 15 m/s
- `strong` if speed &gt; 15 m/s


```python
wspd_morning = 3.2

if wspd_morning < 5:
    print('morning: calm')
elif wspd_morning <= 15:
    print('morning: moderate')
else:
    print('morning: strong')
```

### 1.4 Multi-branch decisions: season from month

Meteorologists often group months into seasons. The same `elif` pattern works for discrete categories:

- Winter: December, January, February (months 12, 1, 2)
- Spring: March–May (3, 4, 5)
- Summer: June–August (6, 7, 8)
- Fall: September–November (9, 10, 11)


<font color="#0F766E">**Task:** `month = 3` is set below. **Do not run the code yet** — predict which season will print. Write your prediction in the next cell.</font>


*(Prediction for month = 3:)*


```python
month = 7

if month in [12, 1, 2]:
    print('Winter')
elif month in [3, 4, 5]:
    print('Spring')
elif month in [6, 7, 8]:
    print('Summer')
elif month in [9, 10, 11]:
    print('Fall')
else:
    print('Invalid month')
```

Notice `month in [6, 7, 8]` — we test whether the value appears **in** a list. Winter months (12, 1, 2) are a special case because the year wraps; we handle that with a list membership test.


### 1.5 Membership: `if x in list`

**Common error:** comparing a whole list to a single number does **not** search inside the list.


```python
sample_readings = [0, 10, 4, 5, 6, 3, 17, 19, 23]
```

<font color="#0F766E">**Task (pair+share):** Look at the code below, but **do not run it yet**. What will print? Write your prediction, then **compare with a neighbor** before we run it together.</font>


*(Prediction for `if sample_readings == 5:`)*


```python
if sample_readings == 5:
    print('The number 5 is in the list')
else:
    print('The number 5 is not in the list')
```

The value 5 **is** in the list, but `sample_readings == 5` compares the **entire list** to the number 5. To test membership, use `in`:


```python
if 5 in sample_readings:
    print('The number 5 is in the list')
```

### 1.6 Compound conditions: `and` and `or`

Sometimes one comparison is not enough:

- **`and`** — both conditions must be `True` (used this afternoon when filtering lidar data by year **and** month)
- **`or`** — at least one condition must be `True` (e.g. any winter month)


```python
year = 2020
month = 6
wind = 18.0

# Preview for Lecture 3: June 2020 records
if year == 2020 and month == 6:
    print('This record is from June 2020')

# Strong wind in June
if month == 6 and wind > 15:
    print('Strong wind in June')
```

Winter months can be written with **`or`**:

`month == 12 or month == 1 or month == 2`

…or with **`in`** (often clearer for a fixed set of values):

`month in [12, 1, 2]`

Both work — pick the style that reads best to you.


```python
month = 1

if month == 12 or month == 1 or month == 2:
    print('Winter (or chain)')

if month in [12, 1, 2]:
    print('Winter (in list)')
```
