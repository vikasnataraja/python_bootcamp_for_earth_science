<h2> 1. Intro to 'if' statements

We can use an 'if' statement to test whether a condition is true and follow a procedure in the event that the condition is true.

```python
# Example of using 'if' statements
a=33
b=33

if b==a:
    print('b equals a')
```

Notice how we use '=' to assign a value to a variable and we use '==' to check if one variable equals another.

We can use 'if', 'elif' [stands for else if], and 'else' statements to test multiple conditions. You can add as many 'elif' statements as needed, but must begin the series of statements with an 'if' statement.

```python
# Example of using 'if' statements considering three cases
a=34
b=33

if a>b:
    print('a is greater than b')
elif a<b:
    print('a is less than b')
else:
    print('a is equal to b')
```

**Common Error**

Sometimes, when using an if statement on an entire list you may run into an error...

```python
# Make a list
l = [0, 10, 4, 5, 6, 3, 17, 19, 23]

# Check if any values in the list equal 5
if l == 5:
    print('The number 5 is in the list')
else:
    print('The number 5 is not in the list')
```

Notice how the value 5 is in the list but our if statement did not work! Why could this be....

Turns out, in order to look through all values of a list, you can use an 'if __ in' statement. More information on this can be found [here](https://stackabuse.com/python-check-if-array-or-list-contains-element-or-value/).

```python
# Check if any values in the list equal 5
if 5 in l: 
    print('The number 5 is in the list')
```
