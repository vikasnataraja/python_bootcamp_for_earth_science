<font color='#0F766E'>**1. Wind speed categories**</font>

Given the three scalar wind speeds below, write an `if` / `elif` / `else` block for **each** speed that prints a category:

- `calm` if speed < 5 m/s
- `moderate` if 5 ≤ speed ≤ 15 m/s
- `strong` if speed > 15 m/s


```python
wspd_morning = 3.2
wspd_afternoon = 9.8
wspd_evening = 18.1

if wspd_morning < 5:
    print("morning: calm")
elif wspd_morning <= 15:
    print("morning: moderate")
else:
    print("morning: strong")

if wspd_afternoon < 5:
    print("afternoon: calm")
elif wspd_afternoon <= 15:
    print("afternoon: moderate")
else:
    print("afternoon: strong")

if wspd_evening < 5:
    print("evening: calm")
elif wspd_evening <= 15:
    print("evening: moderate")
else:
    print("evening: strong")
```

<font color='#0F766E'>**2. Season from month number**</font>

Given `month = 7`, use an `elif` chain to print the season name. Use these definitions:

- Winter: December, January, February (months 12, 1, 2)
- Spring: March–May (3, 4, 5)
- Summer: June–August (6, 7, 8)
- Fall: September–November (9, 10, 11)


```python
month = 7

if month in [12, 1, 2]:
    season = "Winter"
elif month in [3, 4, 5]:
    season = "Spring"
elif month in [6, 7, 8]:
    season = "Summer"
elif month in [9, 10, 11]:
    season = "Fall"
else:
    season = "Unknown"
print(f"Month {month} is in {season}")
```

<font color='#0F766E'>**3. Membership with `in`**</font>

Given `target_month = 3`, use `if target_month in [...]` to print whether Boulder spring fieldwork season is favorable.


```python
target_month = 3

if target_month in [3, 4, 5]:
    print("Favorable: spring fieldwork season in Boulder.")
else:
    print("Not spring fieldwork season.")
```
