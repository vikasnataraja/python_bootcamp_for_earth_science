## Section 5 — `while` loops (not covered in lecture)


---
### <span style="color:red">Dive Deeper: `while` loops (optional)

The morning lesson uses **`for` loops** to step through a known sequence (every index in an array, every fruit name, and so on). A **`while` loop** repeats **while a condition stays `True`** — useful when you do not know in advance how many times to repeat.

📖 **Read:** [Python `while` loops](https://docs.python.org/3/tutorial/controlflow.html#while-statements)

**Key idea:** update the variable in your condition inside the loop, or the loop may never end.

**Example — wind speed drops below a threshold:**


````python
# Simulate wind decreasing until it drops below 10 m/s
wind_speed = 18.0
threshold = 10.0

while wind_speed > threshold:
    print(f'Wind still strong: {wind_speed:.1f} m/s')
    wind_speed -= 2.0

print(f'Wind now below threshold: {wind_speed:.1f} m/s')
````
