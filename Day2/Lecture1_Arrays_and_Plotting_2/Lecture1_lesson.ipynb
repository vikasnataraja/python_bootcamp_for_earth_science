## 1. From Lists to NumPy Arrays

During Day 1 you worked with **lists** and got a first look at NumPy (`np.array`, `np.mean`) and Matplotlib (`plt.plot`). Today we focus on **NumPy arrays** for numeric data and a more customizable plotting style.

Why use NumPy arrays instead of lists? With lists you often need to loop over elements one at a time. NumPy arrays lets you apply math to **every element at once** — essential for scientific data.

We import the whole NumPy package (as most in the scientific community do) as a shortened name `np`:
> `import numpy as np`

Reference: [NumPy quickstart](https://numpy.org/doc/stable/user/quickstart.html)

```python
import numpy as np

# Make one list and one array
my_list = [1, 2, 3]
my_arr = np.array([1, 2, 3])

# Print the type of each
print("my_list is type:")
print(type(my_list))

print("my_arr is type:")
print(type(my_arr))

# Print the result of multiplying each by 2
print('list * 2:', my_list * 2)
print('array * 2:', my_arr * 2)
```

This shows that multiply (`*`) means something different for lists than it does for arrays. Lists get duplicated, where arrays do point-by-point multiplication.

For scientific applications we use NumPy to help ensure our numerical and mathematical operations occur properly and using arrays helps us do this.

In Day 1, you used `np.mean()` to easily compute an average across many numbers using arrays. That pattern scales to larger problems  e.g. thousands of weather observations - and using arrays well can help make these larger calculations be done effeciently


### 1.1 Creating arrays


```python
arr = np.array([0, 1, 2, 3, 4])  # from a list
print(arr)
print(type(arr))
print('shape:', arr.shape)
print('dtype:', arr.dtype)
```

Two common ways to make evenly spaced 1D arrays:

| Function | What it does | Example |
|----------|--------------|--------|
| `np.linspace(start, stop, num)` | `num` evenly spaced values **including** `stop` | `np.linspace(0, 10, 5)` → 5 values from 0 to 10 |
| `np.arange(start, stop, step)` | Values from `start` up to (not including) `stop` | `np.arange(0, 10, 2)` → 0, 2, 4, 6, 8 |


```python
evenly = np.linspace(1, 5, 5)
steps = np.arange(1, 6, 1)
print('linspace:', evenly)
print('arange:  ', steps)
```

For 2D arrays (rows and columns), pass a shape tuple to `np.zeros` or `np.ones`:
> `arr = np.zeros((2, 3))`  # 2 rows, 3 columns


```python
arr_2d = np.zeros((2, 3))
print(arr_2d)
print('shape:', arr_2d.shape)
```

### 1.2 Indexing and slicing (1D)


Use `[index]` for a single value and `[start:stop]` for a slice. **Python starts counting at 0.** Negative indices count from the end (`-1` is the last element).


```python
temps_f = np.array([38.0, 41.2, 48.5, 55.1, 63.0, 71.2, 75.8, 74.1, 65.3, 54.0, 43.2, 39.1])
print('January (index 0):', temps_f[0])
print('Slice months 3–5 (Apr–Jun):', temps_f[3:6])
print('Last month:', temps_f[-1])
```

### 1.3 Indexing 2D arrays


For a 2D array use `arr[row, col]` — **square brackets**, not parentheses.


```python
arr_2d = np.zeros((2, 3))
print(arr_2d)
```

```python
arr_2d[0, 1] = 3.14  # row 0, column 1
print(arr_2d)
```

### 1.4 Vectorized math


Apply arithmetic and **NumPy math functions** to every element at once.

**Geoscience example — temperature conversion:** convert monthly °F to °C:
> `temps_c = (temps_f - 32) * 5 / 9`

**Trig example:** functions like `np.sin`, `np.cos`, and `np.tan` also work on whole arrays (angles must be in **radians**). This is useful for waves, seasonal cycles, and geometry:
> `wave = np.sin(angles)`


```python
# Temperature conversion + summary stats
temps_c = (temps_f - 32) * 5 / 9
print('Mean °F:', np.mean(temps_f))
print('Mean °C:', np.mean(temps_c))
print('Sum of °F:', np.sum(temps_f))

# Trig on an array of angles (radians)
angles = np.linspace(0, 2 * np.pi, 12)  # 12 angles, one per month
wave = np.sin(angles)
print('sin(angles):', np.round(wave, 3))
print('Mean of sin wave:', np.round(np.mean(wave), 3))
```

## 2. Plotting with Matplotlib

### 2.1 From Day 1 to the object-oriented style

On Day 1 you used the quick style:
> `plt.plot(x, y)`

Today we use **figure** and **axes** objects for more control:
> `fig, ax = plt.subplots()` then `ax.plot(x, y)`

We will use the **object-oriented** (`ax.*`) method in this notebook.


```python
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 11, 12)  # month indices
fig, ax = plt.subplots()
ax.plot(x, temps_f, marker='o')
ax.set_xlabel('Month index')
ax.set_ylabel('Temperature (°F)')
ax.set_title('Boulder monthly mean temperature (sample)')
```

In Jupyter, the last line may print an object like `[<matplotlib.lines.Line2D …>]`. Add a **semicolon** to the last line to hide it:
> `ax.plot(x, temps_f);`


### 2.2 Labels and line style


Useful `ax` methods:
- `ax.set_xlabel('text')`, `ax.set_ylabel('text')`, `ax.set_title('text')`
- `ax.plot(x, y, color='k', linewidth=2)`

Docs: [matplotlib.pyplot.plot](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)

Plot two series on the same axes and add a legend:

### 2.4 Plotting real data (Boulder monthly temperatures)


We load CSV data the same way as on Day 1. The file `../data/boco_air_temp.csv` has monthly mean air temperatures for Boulder, CO.
