## Review: how to use lists

Lists are used to store multiple variables in a single variable <br>
Empy lists can be initialised by either of the following: <br>
>x = list() <br>
>x=[ ]

Here is a good tutorial for lists if you want more: https://www.w3schools.com/python/python_lists.asp

````python
x = [0,1,2,3,4] # create a list
print(type(x))
````

As we can see, x is a 'list'.<br>
We can get the length of our list with:
> len(x)

````python
print('Our list has', len(x), 'elements') #print length of our list
````

You can easily access items in a list by 'indexing'<br>
__In python, indexing starts at 0.__ <br>
This means that the first item in our list is at index 0 and the second item is at index 1 and so on...

````python
print('The first item in our list is: ', x[0]) #access the first element of our list (index = 0)
````

Lists are also nice because you can easily add items to them or remove items

````python
x.append(5) #adds 5 to the end of the list
print(x)
````

Lists can contain data of different data types:

````python
x2 = ["Hi", 0, True, "no"]
print(x2)
````

````python
x3 = x + x2
print(x3)
````

## 1.9 Plotting Data

Here is the basic syntax for calling a plot:
```python
plt.plot(xdata,ydata)
```
Remember that `plt` is the name we gave to the package `matplotlib.pyplot` when we imported it

<font color='#0F766E'> Try this yourself by running the cell below.


````python
plt.plot(data.Year, data.Average)
````

<font color='#0F766E'> The function plt.plot used above was given two inputs, data.Year and data.Average. Which appears on the x-axis, and which on the y-axis?
