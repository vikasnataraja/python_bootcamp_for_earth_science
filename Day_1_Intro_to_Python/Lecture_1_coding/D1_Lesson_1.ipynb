### 1.0.1 Programming

Programming/Coding is the process of writing instructions for the computer to implement and execute. I think of this as the language of the computer.

Just like human language, it's complicated. If you've done it for a long time, it may seem easy, but that's because you're familiar with the rules and have had a lot of practice. When you don't know the rules and haven't had much practice, it can feel hard or even impossible to understand what's going on in a conversation, or "program" in code-speak. Hopefully, over this week we can help you start the lifelong process of learning how to use computers to help you solve problems in Earth sciences, or wherever else you end up!

To start, let's think about this by comparing this computer language (programming) to our human languages. First, there are the types of words you use, which in both the computer and human realms we call the **language**. In this bootcamp, the human language in use is English and the computer language is Python; other common languages include Java, C++, Matlab, and R. Next, the particular way you string words together on the human side is called grammar and for the computer we call **syntax**. These are the rules that we obey to make sure any person (or computer) that understands our language can understand us. This bootcamp will help you learn these syntax rules for python!

Lastly, understanding a spoken language can be very different from speaking in that language. To listen to and understand English is a  different skill from speaking clearly in English. Similarly, coding involves understanding what you write and program, **AND** it involves being able to share this with other people so they too can understand what you've done. This is where Jupyter Notebooks come in. Jupyter Notebooks are an **editor**, or the piece of software you write your code with, and one of the best ways to share code with other people (or your future self, who may have forgotten what you were doing when you first wrote a program). You can show people how to use code you've made by giving examples and explain to them what different parts mean using "non-code" blocks, like what you're reading here. ### 1.0.2 Python Code

This section will explain some of the basic structure of Python using code examples that you cannot run. You're not expected to understand all of this yet, so don't worry if it isn't all clear! We will have examples like this later, in section 1.2 and future sections, that you can run.


1. **Python processes code top to bottom, left to right.**

This code works.
```python
x = 3
print(x + 2)
```

This code does not.
```python
print(x + 2)
x = 3
```


2. **When writing python code every new idea needs to happen on a new line because Python works top first, bottom last.** 

This code works.
```python
age = 10
name = "Bella"
```

This code doesn't work.
```python
age = 10 name = "Bella"
```


3. **Spaces are optional around math symbols, but they help make things easier to read, so include them!**

This code works, but is hard to read.
```python
x=3
y=2
z=2*x+y-1
```

This code also works and looks much nicer.
```python
x = 3
y = 2
z = (2 * x) + y - 1
```


3. **Python uses tab spacing to distiguish different sections.**

This code works
```python

    print("This student is younger than 10")

    print("This student is older than 10")
```

This code doesn't work.
```python
if age < 10:
print("This student is younger than 10")
```

## 1.3: Variables

To save a value for use later in python, we use something called a variable.

A variable is a name that you as the coder make up and use as a place holder for a value. There are some rules for what you can and can't use for variable names. The two primary rules are that variable names cannot contain spaces, and they cannot start with a number. Some example good variable names are aNumber, day_name, aName, or a_list.

The `=` is an operation that means: store the value on the right side in the name on the left side.

The example below stores the value of 5 in the name `x`

<font color='#0F766E'> Task: Run the next cell block and try adding another operation on x.


````python
x = 5
print("The variable x starts as ", x)
x = x + 3
print("The variable x now has the value", x)
# Your operation here:
````

Naming of variables is a skill worth practicing. In the line above, the name of `x` is not very descriptive and without context it is not clear what it is meant to represent. Below is an example using good and bad names and how it affects the "readability" of your code.

First is a bad example of variable names that is probably quite confusing.

````python
sa = 19
y2g = 3

ga = sa + y2g
print("Graduation age is ", ga)
````

Here is a good example using variable names which helps you and others read your code and understand what is happening.

````python
student_age = 19
years_to_graduation = 3

graduation_age = student_age + years_to_graduation
print("Graduation age is ", graduation_age)
````

An even better example will include comments. Comments are preceded by a `#` and can appear either on their own line or at the end of a line of code. If you are defining a value, try to include units in the comments.

````python
student_age = 19 # age of the student in years
years_to_graduation = 3 # number of years until the student graduates

# Calculate the graduation age in years
graduation_age = student_age + years_to_graduation
print("Graduation age is ", graduation_age) # print the calculated graduation age
````


Comments in the code can be very helpful to enhance your code's readability. A combination of comments and good variable names will ensure your code is understandable for people other than just yourself.

## 1.4 Common "Bugs" when Coding - Syntax vs Semantics

There will be two types of problems you come across as you start developing code. The process of fixing these problems is called **Debugging**. 

One type of problem is called a **syntax problem**. These are similar to grammar problems in English, like the sentence below. 

This, is: sentence wrong?

These problems use valid words, but break the rules of structure in the language and make no sense to someone who speaks it. In Python, syntax problems will throw an error when you try to run them. Usually a quick Google search of the error name can help understand what is going wrong. However, these errors are not always easy to understand. With practice, discussion with other coders, and time working on code you will find that debugging is a skill that you will always be working on!

Next up are the trickier problems called **semantic, or meaning**, problems. In English you may have seen an issue of meaning like the one below where a single comma can dramatically change the meaning of a sentence:

Let's eat, grandpa.

The first example is an invitation to a family member to join at a meal. The second ... well that's the potential loss of a family member and likely jail time. Each of these follows the rules of language, but the same words can have very different meanings with small changes in structure.
